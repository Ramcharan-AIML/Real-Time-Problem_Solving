```python

"""
1. Add Student
2. Mark Attendance
3. View Attendance
4. Exit

Enter choice: 1
Enter student name: Ravi
Student added successfully


Enter student name: Ravi
Enter status: Present
Attendance updated

"""
```

```python

student_list = []

def add_student(name):

    student_data = {
        "name" : name,
        "status" : "Not Marked"
    }

    student_list.append(student_data)
    print("Student Added Successfully")

    print(student_list)

```

```python

def mark_attendance(stud_name):

    found = False

    for mark in student_list:
        if stud_name.lower() == mark["name"].lower():
            print("Mark the Attendance")
            print("1) Present")
            print("2) Absent")

            pre_abs = str(input("Choose the Option :"))
            print("Attendance updated")

            if pre_abs == "1":
                    mark["status"] = "Present"
            elif pre_abs == "2":
                    mark["status"] = "Absent"
            else:
                print("Wrong Option")
            found = True
    if not found:
        print("Name is not in the Records")

# mark_attendance("ram")
# print(student_list)
```

```python

def view_attendance():
    for view in student_list:
        print(view["name"] ,"-" ,view["status"])
# view_attendance()
```

```python

def main():
    while True:
        print("\n1. Add Student")
        print("2. Mark Attendance")
        print("3. View Attendance")
        print("4. Exit \n")

        choice = str(input("Enter the choice "))

        if choice == "1":
            name = str(input("Enter Student name :"))
            add_student(name)

        elif choice =="2":
            stud_name = str(input("Search the Student Name : "))
            mark_attendance(stud_name)
        elif choice =="3":
            view_attendance()
        elif choice == "4":
            break
main()
```
