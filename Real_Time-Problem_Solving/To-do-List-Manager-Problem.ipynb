```python
"""
1. Add Task
2. View Tasks
3. Mark Task Completed
4. Delete Task
5. Exit

Enter choice: 1
Enter task: Learn functions
Task added successfully


Enter task name: Learn functions
Task marked as completed

Enter task name: Learn functions
Task deleted successfully

"""
```

```python
task_list=[]

def add_task(name):
    task_info = {
        "task": name,
        "status" : "Pending"
    }

    task_list.append(task_info)
    print(task_list)
```

```python
def view_task():
    for view_task in task_list:
        print(f"{view_task["task"]} - {view_task["status"]}")
```

```python
def mark_task(mark_name):

    found = False

    for mark_task in task_list:
        if mark_name.lower() == mark_task["task"].lower():
            found = True

            print(f"Press 1 to mark the {mark_task["task"]} task Completed")
            print("1) Completed")
            option = str(input("Press 1 : "))

            if option == "1":
                mark_task["status"] = "Completed"
                
                print("Task marked as completed ✅")
            else:
                print("Wrong Option ❌")
                

    if not found:
        print("Task not Found in the List")

# mark_task("brushing")
```

```python
def delete_task(track_name):

    found = False

    for del_task in task_list:
        if track_name.lower() == del_task["task"].lower():
            found = True

            print(f"Press 1 to delete the {del_task["task"]} task")
            print("1) Delete")
            option = str(input("Press 1 : "))

            if option == "1":
                task_list.remove(del_task)
                print("Task Deleted Succesfully ✅")
            else:
                print("Wrong Option ❌")
                

    if not found:
        print("Task not Found in the List❌")

delete_task("breakfast")
```

```python
def main():
    while True:
        print("\n1. Add Task")
        print("2. View Tasks")
        print("3. Mark Task Completed")
        print("4. Delete Task")
        print("5. Exit\n")

        choice = str(input("Enter the Choice : "))

        if choice == "1":
            name = str(input("Enter the name : "))
            add_task(name)
        elif choice == "2":
            view_task()
        elif choice == "3":
            mark_name = str(input("Enter the Task name : "))
            mark_task(mark_name)
        elif choice == "4":
            track_name = str(input("Enter the Task name :"))
            delete_task(track_name)
        elif choice == "5":
            break
        else:
            print("Wrong Option ❌")
        
main()
```
